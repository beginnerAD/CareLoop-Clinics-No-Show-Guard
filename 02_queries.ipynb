# Part 2 — SQL: verify the numbers and find where no-shows cluster

```python
# Task 2.2 Python check: Reproduce clinic counts and no-show counts with groupby
clinic_summary = (
    df_clean.merge(
        clinics_df[["clinic_id", "clinic_name", "city"]], on="clinic_id"
    )
    .groupby(["clinic_id", "clinic_name", "city"], as_index=False)
    .agg(appointments=("no_show", "count"), no_shows=("no_show", "sum"))
)
clinic_summary["no_show_pct"] = (
    100.0 * clinic_summary["no_shows"] / clinic_summary["appointments"]
).round(1)
clinic_summary = clinic_summary.sort_values("no_show_pct", ascending=False)
print(clinic_summary.to_string(index=False))
```
